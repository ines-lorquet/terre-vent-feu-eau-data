```python
import pandas as pd

mouvements = pd.read_csv(
    "data/raw/v_mvt_commune_2026.csv",
    dtype=str
)

print(mouvements.shape)
print(mouvements.columns.tolist())
```

```python
# VERIFICATION DES CODES INSEE NON APPARIES DANS L'HISTORIQUE INSEE

mouvements["COM_AV"] = mouvements["COM_AV"].str.strip()
mouvements["COM_AP"] = mouvements["COM_AP"].str.strip()

sans_correspondance["Code INSEE"] = (
    sans_correspondance["Code INSEE"]
    .astype(str)
    .str.strip()
)

codes_non_apparies = (
    sans_correspondance["Code INSEE"]
    .dropna()
    .unique()
)

mouvements_trouves = mouvements[
    mouvements["COM_AV"].isin(codes_non_apparies)
].copy()

print(
    "Nombre de codes non appariés retrouvés dans l'historique INSEE :",
    mouvements_trouves["COM_AV"].nunique()
)

print(
    "Nombre total de codes non appariés :",
    len(codes_non_apparies)
)
```

```python
# IDENTIFICATION DES CODES NON EXPLIQUES PAR L'HISTORIQUE INSEE

codes_retrouves = set(mouvements_trouves["COM_AV"].unique())

codes_non_expliques = [
    code for code in codes_non_apparies
    if code not in codes_retrouves
]

print("Codes restant à expliquer :", codes_non_expliques)

display(
    sans_correspondance[
        sans_correspondance["Code INSEE"].isin(codes_non_expliques)
    ][
        ["Code INSEE", "Nom de la commune", "Département", "Année"]
    ]
    .drop_duplicates()
    .sort_values(["Code INSEE", "Année"])
)
```

```python
# NOMBRE D'INCENDIES EXPLIQUES PAR LES CHANGEMENTS ADMINISTRATIFS

lignes_expliquees = sans_correspondance[
    sans_correspondance["Code INSEE"].isin(codes_retrouves)
]

print(
    "Incendies associés à un ancien code retrouvé :",
    len(lignes_expliquees)
)

print(
    "Incendies restant sans explication :",
    len(sans_correspondance) - len(lignes_expliquees)
)
```

```python
# RATTACHEMENT DES ANCIENS CODES INSEE AUX COMMUNES ACTUELLES

# Nettoyage des codes INSEE

communes["code_insee"] = (
    communes["code_insee"]
    .astype(str)
    .str.strip()
)

mouvements["COM_AV"] = (
    mouvements["COM_AV"]
    .astype(str)
    .str.strip()
)

mouvements["COM_AP"] = (
    mouvements["COM_AP"]
    .astype(str)
    .str.strip()
)

sans_correspondance["Code INSEE"] = (
    sans_correspondance["Code INSEE"]
    .astype(str)
    .str.strip()
)


# Ensemble des codes INSEE présents dans le référentiel communal actuel

codes_actuels = set(
    communes["code_insee"]
    .dropna()
    .unique()
)


# Liste des codes INSEE BDIFF qui n'ont pas trouvé de correspondance

codes_non_apparies = (
    sans_correspondance["Code INSEE"]
    .dropna()
    .unique()
)


# Table simplifiée des changements de codes INSEE

passages = (
    mouvements[
        ["COM_AV", "COM_AP"]
    ]
    .dropna()
    .drop_duplicates()
)


# Vérification des anciens codes qui ont plusieurs codes après événement

plusieurs_destinations = (
    passages
    .groupby("COM_AV")["COM_AP"]
    .nunique()
)

plusieurs_destinations = plusieurs_destinations[
    plusieurs_destinations > 1
]

print(
    "Nombre d'anciens codes ayant plusieurs destinations :",
    len(plusieurs_destinations)
)


# Création du dictionnaire ancien code -> nouveau code

# Attention :
# on utilise ici uniquement les anciens codes qui ont une seule destination.
# Les cas complexes seront vérifiés séparément.

passages_uniques = (
    passages[
        passages["COM_AV"].isin(
            plusieurs_destinations.index
        ) == False
    ]
)

passage_dict = dict(
    zip(
        passages_uniques["COM_AV"],
        passages_uniques["COM_AP"]
    )
)


# Fonction permettant de suivre les changements administratifs
# jusqu'à retrouver un code présent dans le référentiel actuel

def trouver_code_actuel(code, max_etapes=20):

    code = str(code).strip()

    historique = [code]

    for _ in range(max_etapes):

        # Si le code existe dans le référentiel actuel,
        # on considère que la commune actuelle est retrouvée

        if code in codes_actuels:
            return code, historique, "retrouve"

        # Si le code a plusieurs destinations possibles,
        # on classe le cas comme complexe

        if code in plusieurs_destinations.index:
            return None, historique, "cas_complexe"

        # Si aucun changement n'est trouvé dans l'historique INSEE

        if code not in passage_dict:
            return None, historique, "non_retrouve"

        nouveau_code = passage_dict[code]

        # Protection contre une éventuelle boucle

        if nouveau_code in historique:
            return None, historique, "boucle"

        code = nouveau_code
        historique.append(code)

    return None, historique, "max_etapes"


# Application aux codes BDIFF non appariés

resultats_codes = []

for code in codes_non_apparies:

    code_actuel, historique, statut = trouver_code_actuel(code)

    resultats_codes.append(
        {
            "code_insee_bdiff": code,
            "code_insee_actuel": code_actuel,
            "statut": statut,
            "chaine_codes": " -> ".join(historique)
        }
    )


correspondance_codes = pd.DataFrame(resultats_codes)


# Ajout du nom de la commune actuelle

correspondance_codes = correspondance_codes.merge(
    communes[
        [
            "code_insee",
            "nom_standard"
        ]
    ],
    left_on="code_insee_actuel",
    right_on="code_insee",
    how="left"
)


# Suppression de la colonne technique issue de la jointure

correspondance_codes = correspondance_codes.drop(
    columns=["code_insee"],
    errors="ignore"
)


# Renommage pour plus de clarté

correspondance_codes = correspondance_codes.rename(
    columns={
        "nom_standard": "nom_commune_actuelle"
    }
)


# Résumé des résultats

print(
    "Nombre total de codes non appariés :",
    len(correspondance_codes)
)

print(
    "Codes rattachés à une commune actuelle :",
    (
        correspondance_codes["statut"] == "retrouve"
    ).sum()
)

print(
    "Cas complexes :",
    (
        correspondance_codes["statut"] == "cas_complexe"
    ).sum()
)

print(
    "Codes non retrouvés :",
    (
        correspondance_codes["statut"] == "non_retrouve"
    ).sum()
)


# Affichage des correspondances

display(
    correspondance_codes[
        [
            "code_insee_bdiff",
            "code_insee_actuel",
            "nom_commune_actuelle",
            "statut",
            "chaine_codes"
        ]
    ]
    .sort_values(
        [
            "statut",
            "code_insee_bdiff"
        ]
    )
)


# Vérification spécifique des cas non rattachés

display(
    correspondance_codes[
        correspondance_codes["statut"] != "retrouve"
    ][
        [
            "code_insee_bdiff",
            "statut",
            "chaine_codes"
        ]
    ]
)
```

```python
# CONSTRUCTION D'UNE CHAINE HISTORIQUE VERS UN CODE ACTUEL

# Préparation des dates

mouvements["DATE_EFF"] = pd.to_datetime(
    mouvements["DATE_EFF"],
    errors="coerce"
)

# On garde uniquement les lignes où le code change réellement

transitions = mouvements[
    mouvements["COM_AV"] != mouvements["COM_AP"]
].copy()

transitions = transitions.sort_values(
    ["COM_AV", "DATE_EFF"]
)


# Fonction de recherche du code actuel

def trouver_code_actuel_chrono(code, max_etapes=20):

    code_courant = str(code).strip()
    historique = [code_courant]

    for _ in range(max_etapes):

        # Si le code existe aujourd'hui
        if code_courant in codes_actuels:
            return code_courant, historique, "retrouve"

        # Toutes les transitions connues pour ce code
        candidats = transitions[
            transitions["COM_AV"] == code_courant
        ].copy()

        # Aucun événement trouvé
        if candidats.empty:
            return None, historique, "non_retrouve"

        # On prend la transition la plus récente
        candidats = candidats.sort_values(
            "DATE_EFF",
            ascending=False
        )

        nouveau_code = candidats.iloc[0]["COM_AP"]

        # Protection contre les boucles
        if nouveau_code in historique:
            return None, historique, "boucle"

        code_courant = nouveau_code
        historique.append(code_courant)

    return None, historique, "max_etapes"


# Application aux codes non appariés

resultats_codes = []

for code in codes_non_apparies:

    code_actuel, historique, statut = trouver_code_actuel_chrono(code)

    resultats_codes.append({
        "code_insee_bdiff": code,
        "code_insee_actuel": code_actuel,
        "statut": statut,
        "chaine_codes": " -> ".join(historique)
    })


correspondance_codes = pd.DataFrame(resultats_codes)


# Ajout du nom actuel

correspondance_codes = correspondance_codes.merge(
    communes[
        ["code_insee", "nom_standard"]
    ],
    left_on="code_insee_actuel",
    right_on="code_insee",
    how="left"
)

correspondance_codes = correspondance_codes.rename(
    columns={
        "nom_standard": "nom_commune_actuelle"
    }
)

correspondance_codes = correspondance_codes.drop(
    columns=["code_insee"],
    errors="ignore"
)


# Résumé

print(
    "Nombre total de codes non appariés :",
    len(correspondance_codes)
)

print(
    "Codes rattachés à une commune actuelle :",
    (correspondance_codes["statut"] == "retrouve").sum()
)

print(
    "Codes non retrouvés :",
    (correspondance_codes["statut"] == "non_retrouve").sum()
)

print(
    "Boucles détectées :",
    (correspondance_codes["statut"] == "boucle").sum()
)


# Affichage

display(
    correspondance_codes[
        [
            "code_insee_bdiff",
            "code_insee_actuel",
            "nom_commune_actuelle",
            "statut",
            "chaine_codes"
        ]
    ]
)
```
